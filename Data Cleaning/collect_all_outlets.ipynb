```python
import pandas as pd
```

```python
elite_df = pd.read_csv('Elite_Predictions_2026_01.csv')
high_df    = pd.read_csv('High_Predictions_2026_01.csv')
medium_df  = pd.read_csv('Medium_Predictions_2026_01.csv')
low_df     = pd.read_csv('Low_Predictions_2026_01.csv')
```

```python
combined_df = pd.concat([elite_df, high_df, medium_df, low_df], ignore_index=True)

# Sort by Outlet_ID ascending
combined_df = combined_df.sort_values('Outlet_ID').reset_index(drop=True)

# Save
combined_df.to_csv('All_Predictions_2026_01.csv', index=False)
```
