```python
import pandas as pd
distributorDF = pd.read_csv('Bronze/distributor_seasonality_details.csv')
```

```python
distributorDF.info()
```

```python
distributorDF['year_month'] = distributorDF['Year'].astype(str) + '-' + distributorDF['Month'].astype(str).str.zfill(2)
```

```python
distributorDF.head(10)
```

```python
distributorDF = distributorDF.sort_values('year_month') 
```

```python
distributorDF.head(10)
```

```python
distributorDF = distributorDF.drop(columns=['Year'])

```

```python
distributorDF = distributorDF.drop(columns=['Month'])
```

```python
cols = ['year_month', 'Distributor_ID', 'Seasonality_Index'] + [col for col in distributorDF.columns if col not in ['year_month', 'Distributor_ID', 'Seasonality_Index']]
```

```python
distributorDF = distributorDF[cols]
```

```python
distributorDF.to_csv(r'Silver/distributor_seasonality_details_by_clean.csv',index=False)
```

```python
trnasactionHistoryDF = pd.read_csv('Bronze/transactions_history_final.csv')
```

```python
trnasactionHistoryDF.head(10)
```

```python
trnasactionHistoryDF = trnasactionHistoryDF.drop(columns='Total_Bill_Value')
```

```python
trnasactionHistoryDF = trnasactionHistoryDF.drop(columns='SKU_ID')
```

```python
trnasactionHistoryDF['year_month'] = trnasactionHistoryDF['Year'].astype(str) + '-' + trnasactionHistoryDF['Month'].astype(str)
```

```python
trnasactionHistoryDF = trnasactionHistoryDF.drop(columns='Year')
```

```python
trnasactionHistoryDF = trnasactionHistoryDF.drop(columns='Month')
```

```python
trnasactionHistoryDF.head(10)
```

```python
cols1 = ['Outlet_ID', 'year_month', 'Distributor_ID','Volume_Liters'] + [col for col in trnasactionHistoryDF.columns if col not in ['Outlet_ID', 'year_month', 'Distributor_ID','Volume_Liters']]
```

```python
trnasactionHistoryDF = trnasactionHistoryDF[cols1]
```

```python
trnasactionHistoryDF.head(10)
```

```python
trnasactionHistoryDF['year_month'] = pd.to_datetime(trnasactionHistoryDF['year_month'],format='%Y-%m')
```

```python
trnasactionHistoryDF = trnasactionHistoryDF.sort_values(by=['Outlet_ID','year_month']).reset_index(drop = True)
```

```python
trnasactionHistoryDF['year_month'] = trnasactionHistoryDF['year_month'].dt.strftime('%Y-%m')
```

```python
trnasactionHistoryDF.head(20)
```

```python
trnasactionHistoryDF.to_csv(r'Silver/transactions_history_final_by_clean.csv',index=False)
```

```python
trnasactionHistoryDF1 = trnasactionHistoryDF.drop(columns='Distributor_ID')
```

```python
trnasactionHistoryDF1.head(10)
```

```python
trnasactionHistoryDF1 = trnasactionHistoryDF1.groupby(['Outlet_ID','year_month'])['Volume_Liters'].sum().reset_index()
```

```python
trnasactionHistoryDF1 = trnasactionHistoryDF1.sort_values(by=['Outlet_ID','year_month'])
```

```python
trnasactionHistoryDF1 = trnasactionHistoryDF1.rename(
    columns={
        'Volume_Liters':'Volume_Liters_Sum'
    }
    )
```

```python
trnasactionHistoryDF1.head(20)
```

```python
trnasactionHistoryDF1.to_csv(r'Silver/transactions_history_final_by_clean_totalSum_monthly.csv',index=False)
```

```python
# HOLIDAY COUNTS
# ----------------------------------------------------------------------------
```

```python
holidayCountDF = pd.read_csv('Silver/holiday_list_by_clean.csv')
```

```python
holidayCountDF.head(10)
```

```python
holidayCountDF = holidayCountDF.sort_values('Date')
```

```python
holidayCountDF.head(10)
```

```python
holidayCountDF.to_csv(r'Silver/holiday_list_by_sort.csv',index=False)
```

```python
holidayCountDF['Date'] = pd.to_datetime(holidayCountDF['Date'])
```

```python
# holidayCountDF['year_month'] = holidayCountDF['Date'].dt.strftime('%Y-%m-%d')
```

```python
holidayCountDF = holidayCountDF.groupby('Date')['Date'].nunique().reset_index(name='monthly_holiday_counts')
```

```python
holidayCountDF.head(10)
```

```python
holidayCountDF['year_month'] = holidayCountDF['Date'].dt.strftime('%Y-%m')
```

```python
holidayCountDF = holidayCountDF.groupby('year_month')['Date'].nunique().reset_index(name='monthly_holiday_count')
```

```python
holidayCountDF = holidayCountDF.sort_values('year_month').reset_index(drop=True)
```

```python
holidayCountDF.head(10)
```

```python
holidayCountDF.to_csv(r'Silver/holiday_list_by_clean.csv',index=False)
```
