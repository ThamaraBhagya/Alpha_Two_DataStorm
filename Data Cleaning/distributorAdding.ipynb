```python
import pandas as pd
outlet_dist_df  = pd.read_csv('Silver/outlet_distributor.csv')
seasonality_df   = pd.read_csv('Silver/distributor_seasonality_details_by_clean.csv')
elite_df  = pd.read_csv('Gold/Elite_Potential_Hubs_Monthly_Volume.csv')
high_df  = pd.read_csv('Gold/High_Potential_Hubs_Monthly_Volume.csv')
low_df  = pd.read_csv('Gold/Low_Potential_Hubs_Monthly_Volume.csv')
medium_df  = pd.read_csv('Gold/Medium_Potential_Hubs_Monthly_Volume.csv')
```

```python
elite_df = elite_df.merge(outlet_dist_df, on='Outlet_ID', how='left')
high_df = high_df.merge(outlet_dist_df, on='Outlet_ID', how='left')
medium_df = medium_df.merge(outlet_dist_df, on='Outlet_ID', how='left')
low_df = low_df.merge(outlet_dist_df, on='Outlet_ID', how='left')
```

```python
elite_df = elite_df.merge(seasonality_df, on=['Distributor_ID', 'year_month'], how='left')
high_df = high_df.merge(seasonality_df, on=['Distributor_ID', 'year_month'], how='left')
medium_df = medium_df.merge(seasonality_df, on=['Distributor_ID', 'year_month'], how='left')
low_df = low_df.merge(seasonality_df, on=['Distributor_ID', 'year_month'], how='left')
```
