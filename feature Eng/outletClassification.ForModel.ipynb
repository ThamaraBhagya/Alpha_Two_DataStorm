```python
import pandas as pd
lowPotentialHubDF = pd.read_csv('Silver/Low_Potential_Hubs.csv')
monthlyVolunmDF = pd.read_csv('Silver/transactions_history_final_by_clean_totalSum_monthly.csv')
```

```python
lowPotentialHubDF.head(10)
```

```python
monthlyVolunmDF.head(10)
```

```python
redultDF = monthlyVolunmDF[monthlyVolunmDF['Outlet_ID'].isin(lowPotentialHubDF['Outlet_ID'])]
```

```python
redultDF['year_month'] = pd.to_datetime(redultDF['year_month'], format='%Y-%m')
redultDF = redultDF.sort_values(by=['Outlet_ID', 'year_month']).reset_index(drop=True)
redultDF['year_month'] = redultDF['year_month'].dt.strftime('%Y-%m')
```

```python
import pandas as pd
mediumPotentialHubDF = pd.read_csv('Silver/Medium_Potential_Hubs.csv')
highPotentialHubDF = pd.read_csv('Silver/High_Potential_Hubs.csv')
elitePotentialHubDF = pd.read_csv('Silver/Elite_Potential_Hub.csv')
```

```python
redultDF2 = monthlyVolunmDF[monthlyVolunmDF['Outlet_ID'].isin(mediumPotentialHubDF['Outlet_ID'])]
redultDF3 = monthlyVolunmDF[monthlyVolunmDF['Outlet_ID'].isin(highPotentialHubDF['Outlet_ID'])]
redultDF4 = monthlyVolunmDF[monthlyVolunmDF['Outlet_ID'].isin(elitePotentialHubDF['Outlet_ID'])]
```

```python
redultDF2['year_month'] = pd.to_datetime(redultDF2['year_month'], format='%Y-%m')
redultDF2 = redultDF2.sort_values(by=['Outlet_ID', 'year_month']).reset_index(drop=True)
redultDF2['year_month'] = redultDF2['year_month'].dt.strftime('%Y-%m')
```

```python
redultDF3['year_month'] = pd.to_datetime(redultDF3['year_month'], format='%Y-%m')
redultDF3 = redultDF3.sort_values(by=['Outlet_ID', 'year_month']).reset_index(drop=True)
redultDF3['year_month'] = redultDF3['year_month'].dt.strftime('%Y-%m')
```

```python
redultDF4['year_month'] = pd.to_datetime(redultDF4['year_month'], format='%Y-%m')
redultDF4 = redultDF4.sort_values(by=['Outlet_ID', 'year_month']).reset_index(drop=True)
redultDF4['year_month'] = redultDF4['year_month'].dt.strftime('%Y-%m')
```

```python
holidaySortDF = pd.read_csv('Silver/holiday_list_by_clean.csv')
outletDetailDF = pd.read_csv('Bronze/outlet_master.csv')
transactionWithDistributorDF = pd.read_csv('Silver/transactions_history_final_by_clean.csv')
distributorSeasonDF = pd.read_csv('Silver/distributor_seasonality_details_by_clean.csv')
```

```python

redultDF1_2 = redultDF.merge(holidaySortDF, on='year_month', how='left')
resultDF2_2 = redultDF2.merge(holidaySortDF, on='year_month', how='left')
resultDF3_2 = redultDF3.merge(holidaySortDF, on='year_month', how='left')
resultDF4_2 = redultDF4.merge(holidaySortDF, on='year_month', how='left')
```

```python
redultDF1_2 = redultDF1_2.merge(outletDetailDF, on='Outlet_ID', how='left')
resultDF2_2 = resultDF2_2.merge(outletDetailDF, on='Outlet_ID', how='left')
resultDF3_2 = resultDF3_2.merge(outletDetailDF, on='Outlet_ID', how='left')
resultDF4_2 = resultDF4_2.merge(outletDetailDF, on='Outlet_ID', how='left')
```
