```python
import pandas as pd
outletSumDF = pd.read_csv('Silver/transactions_history_final_by_clean_totalSum.csv')
```

```python
outletSumDF = outletSumDF.rename(
    columns={
        'Volume_Liters_Sum_total':'Outlet_ID'
    }
)
```

```python
outletSumDF.head(10)
```

```python
outletSumDF = outletSumDF.groupby(['Outlet_ID']).sum().reset_index(names='Volume_Liters_Sum_total')
```

```python
outletSumDF.head(10)
```

```python
outletSumDF = outletSumDF.sort_values('Year').reset_index(drop=True)
```

```python
outletSumDF.to_csv(r'Silver/transactions_history_final_by_clean_totalSum.csv',index=False)
```

```python
import matplotlib.pyplot as plt
```

```python
outletSumDF.columns = outletSumDF.columns.str.strip()
```

```python


plt.figure(figsize=(12, 6))
plt.bar(outletSumDF['Outlet_ID'], outletSumDF['Volume_Liters_Sum'], color='steelblue')

plt.title('Volume Liters Sum by Outlet ID')
plt.xlabel('Outlet ID')
plt.ylabel('Volume Liters Sum')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
